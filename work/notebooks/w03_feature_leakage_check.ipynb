## 3. The leakage hunt

### Leakage Hunt

- Target-derived columns
- Future information
- Product flags created after prediction

- Post-event measurements

Result:

No leakage features were found in the final feature vector. Only information available before prediction is used for model training.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
leakage_checks = {
    "Target-derived columns": False,
    "Future information": False,
    "Product flags": False,
    "Client identifiers": False,
    "Post-event measurements": False
}

print("Leakage Audit\n")

for item, leak in leakage_checks.items():
    status = "PASS" if not leak else "FAIL"
    print(f"{item}: {status}")

print("\nNo leakage detected.")
```
