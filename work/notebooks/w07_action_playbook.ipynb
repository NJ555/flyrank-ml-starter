## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*
## Ranked content actions

### 1. Review First
Pages with low CTR, poor average position, and high impressions should be reviewed first because they offer the greatest opportunity for improvement.

### 2. Review Next
Pages with moderate traffic or declining freshness should be reviewed after the highest-priority pages.

### 3. Monitor
Pages performing well should be monitored regularly but do not require immediate action.

### Reason codes
- low_ctr
- poor_position
- stale_content
- declining_clicks
- general_review

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
actions = [
    ("Review First", "Low CTR + High Impressions"),
    ("Review Next", "Declining freshness"),
    ("Monitor", "Healthy performance")
]

for rank, (action, reason) in enumerate(actions, start=1):
    print(f"{rank}. {action} -> {reason}")
```
