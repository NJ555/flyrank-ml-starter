## My lane as an ML task

My lane is customer review sentiment analysis.

This is a **classification** problem because each customer review belongs to one of several predefined categories such as Positive, Neutral, or Negative.

The model learns patterns from past reviews and predicts the sentiment of new reviews automatically.

```python
task_type = "Classification"

print("Task Type:", task_type)
```

## Target or Proxy

The target is the sentiment of each customer review.

Possible labels:

- Positive
- Neutral
- Negative

These labels come from previously labeled customer reviews that are used to train the model.

```python
target = ["Positive", "Neutral", "Negative"]

print(target)
```

## Success Metric

The primary evaluation metric is Accuracy.

A model with an accuracy above 90% would generally be considered good for this task.

If the dataset becomes imbalanced, F1 Score can also be used because it balances precision and recall.

```python
accuracy = 0.92

print("Example Accuracy:", accuracy)
```

## Unit of Analysis

Each row represents one customer review.

The dataset contains:

- Review Text
- Rating
- Sentiment

Every row is an independent example used for training the machine learning model.

```python
import pandas as pd

df = pd.DataFrame({
    "Review": [
        "Amazing product",
        "Very bad quality",
        "Average performance",
        "Excellent phone",
        "Poor battery"
    ],
    "Rating":[5,1,3,5,2],
    "Sentiment":[
        "Positive",
        "Negative",
        "Neutral",
        "Positive",
        "Negative"
    ]
})

df
```

## Why ML beats a fixed rule here

A fixed rule such as checking whether a review contains the word "good" is not reliable.

For example:

- "Not good" is actually negative.
- "Pretty decent" is positive even though it does not contain "good".

Machine learning learns these patterns from data instead of relying on manually written rules.

Therefore, ML provides much better predictions for real-world customer reviews.

```python
examples = [
    "Good phone",
    "Not good phone",
    "Pretty decent phone"
]

for sentence in examples:
    print(sentence)
```
