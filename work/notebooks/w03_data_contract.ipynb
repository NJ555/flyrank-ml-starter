```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from datasets import load_dataset
import pandas as pd
from itertools import islice

ds = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    streaming=True,
    split="train"
)

sample = pd.DataFrame(list(islice(ds, 5000)))

print("1) Grain check")
print("rows:", len(sample))
print("duplicate client-content-date rows:", sample.duplicated(["client_hash_id", "content_hash_id", "report_date"]).sum())

print("\n2) Date window")
print("min date:", sample["report_date"].min())
print("max date:", sample["report_date"].max())
print("unique report dates:", sample["report_date"].nunique())

print("\n3) Missing / availability")
check_cols = ["gsc_data_available", "ga4_data_available", "gsc_impressions", "gsc_clicks", "ga4_pageviews", "ga4_sessions"]
print(sample[check_cols].isna().sum().to_dict())
print(sample[["gsc_data_available", "ga4_data_available"]].value_counts(dropna=False).head())
```
